```python

import numpy as np

# given data points
x = np.array([0, 1, 2, 3, 4], dtype=float)
y = np.array([1, 2, 5, 10, 17], dtype=float)

xp = 2.5  # point to interpolate
n = len(x)
h = np.diff(x)

# initialize the system of equations
A = np.zeros((n, n))
b = np.zeros(n)

# natural boundary conditions
A[0, 0] = 1
A[-1, -1] = 1

# construct the tridiagonal system
for i in range(1, n - 1):
    A[i, i - 1] = h[i - 1]
    A[i, i] = 2 * (h[i - 1] + h[i])
    A[i, i + 1] = h[i]

    b[i] = 6 * ((y[i + 1] - y[i]) / h[i]
                - (y[i] - y[i - 1]) / h[i - 1])

# solve for second derivatives
M = np.linalg.solve(A, b)

# find the interpolation interval
i = np.searchsorted(x, xp) - 1
i = np.clip(i, 0, n - 2)

# calculate the interpolated value
a = (x[i + 1] - xp) / h[i]
c = (xp - x[i]) / h[i]

result = (a * y[i] + c * y[i + 1]
          + ((a**3 - a) * M[i]
          + (c**3 - c) * M[i + 1]) * h[i]**2 / 6)

# display the result
print(f"Interpolated value at x = {xp}: {result:.6f}")
```
