```python

import numpy as np
import matplotlib.pyplot as plt

# given data points
x = np.array([1, 2, 4, 7], dtype=float)
y = np.array([1, 4, 16, 49], dtype=float)

xp = 3.5  # point to interpolate
n = len(x)

# Lagrange interpolation function
def lagrange(xp):
    result = 0

    for i in range(n):
        term = y[i]

        for j in range(n):
            if i != j:
                term *= (xp - x[j]) / (x[i] - x[j])

        result += term

    return result

# calculate the interpolated value
yp = lagrange(xp)

print(f"Interpolated value at x = {xp} is {yp:.6f}\n")

# generate points for a smooth interpolation curve
x_curve = np.linspace(min(x), max(x), 300)
y_curve = [lagrange(xi) for xi in x_curve]

# plot the interpolation curve and data points
plt.figure(figsize=(9, 5))

plt.plot(x_curve, y_curve, 'b-', linewidth=2, label='Lagrange Interpolation')
plt.scatter(x, y, color='green', s=60, zorder=4, label='Given Data Points')
plt.scatter(xp, yp, color='red', s=90, zorder=4, label='Interpolated Point')

plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Lagrange Interpolation')
plt.legend()
plt.grid(True, linestyle='-', alpha= 1.0)
plt.tight_layout()
plt.show()
```
