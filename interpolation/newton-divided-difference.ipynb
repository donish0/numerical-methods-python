```python

import numpy as np

# given data points
x = np.array([1, 2, 4, 7], dtype=float)
y = np.array([1, 4, 16, 49], dtype=float)

xp = 3.5  # Point to interpolate
n = len(x)

# initialize the divided difference table
diff = np.zeros((n, n))
diff[:, 0] = y

# calculate divided differences
for j in range(1, n):
    for i in range(n - j):
        diff[i, j] = (diff[i + 1, j - 1] - diff[i, j - 1]) / (x[i + j] - x[i])

# calculate the interpolation value
result = diff[0, 0]
term = 1

for i in range(1, n):
    term *= (xp - x[i - 1])
    result += term * diff[0, i]

# display the result
print(f"Interpolated value at x = {xp}: {result:.6f}")
```
