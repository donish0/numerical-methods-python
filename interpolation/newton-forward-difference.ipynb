```python

import numpy as np

# given data points
x = np.array([0, 1, 2, 3, 4], dtype=float)
y = np.array([1, 2, 5, 10, 17], dtype=float)

xp = 2.5  # point to interpolate
n = len(x)
h = x[1] - x[0]

# initialize the forward difference table
diff = np.zeros((n, n))
diff[:, 0] = y

# calculate forward differences
for j in range(1, n):
    for i in range(n - j):
        diff[i, j] = diff[i + 1, j - 1] - diff[i, j - 1]

# calculate the interpolation value
p = (xp - x[0]) / h
result = y[0]
term = 1

for i in range(1, n):
    term *= (p - i + 1) / i
    result += term * diff[0, i]

# display the result
print(f"Interpolated value at x = {xp}: {result:.6f}")
```
