```python

import sympy as sp

# Gauss-Legendre quadrature method
def GaussLegendre(f, a, b, n):

    x = sp.symbols('x')
    pn = sp.legendre(n, x)

    # find and sort the roots
    xi = sorted([float(root) for root in sp.nroots(pn)])

    # calculate weights
    wi = []

    for i in range(n):

        weight = 2 / (
            (1 - xi[i]**2) *
            (float(sp.diff(pn, x).subs(x, xi[i]))**2)
        )

        wi.append(weight)

    # apply Gauss-Legendre formula
    result = (b - a) / 2 * sum(
        wi[i] * f((b - a) / 2 * xi[i] + (a + b) / 2)
        for i in range(n)
    )

    return result


# define the integrand
def f(x):
    return 1 / (1 + x**2)


# enter the number of integration points
n = int(input("Enter number of points (1, 2, 3, ...): "))
print("")

# display the selected method
if n == 1:
    print("1-Point Gauss-Legendre Integration.")

elif n == 2:
    print("2-Point Gauss-Legendre Integration.")

elif n == 3:
    print("3-Point Gauss-Legendre Integration.")

else:
    print(f"{n}-Point Gauss-Legendre Integration.")

# Calculate and display the result
result = GaussLegendre(f, 0, 1, n)

print("Result:", result)
```
