##### Example :$$f(x) = x^2e^{-x} + \sin(x)$$

```python

import numpy as np

# define the function
def f(x):
    return x**2 * np.exp(-x) + np.sin(x)

# set integration limits and iterations
a = 0
b = 1
n = 4

# initialize the Romberg matrix
R = np.zeros((n, n))

for i in range(1, n + 1):
    m = 2**(i - 1)          # number of subintervals
    h = (b - a) / m         # step size

    # trapezoidal rule approximation
    R[i-1, 0] = (h/2) * (f(a) + 2*sum(f(a + k*h) for k in range(1, m)) + f(b))

    # richardson extrapolation
    for j in range(2, i + 1):
        R[i-1, j-1] = R[i-1, j-2] + (R[i-1, j-2] - R[i-2, j-2]) / (4**(j-1) - 1)

# display the final result and Romberg matrix
print("Romberg integral:", R[n-1, n-1])
print("\nMatrix Form:\n")
print(R)
```
