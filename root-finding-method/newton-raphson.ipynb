```python
import pandas as pd

def newton_raphson(f, df, x0, tol=0.0001, max_iter=100):

    data = []

    for i in range(1, max_iter + 1):

        x1 = x0 - f(x0) / df(x0)

        error = abs(x1 - x0)

        data.append([i, x0, f(x0), x1, error])

        if error < tol:
            break

        x0 = x1

    table = pd.DataFrame(data, columns=[
        "Iteration", "x0", "f(x0)", "x1", "Error"
    ])

    return table, x1


f = lambda x: x**3 - x - 2
df = lambda x: 3*x**2 - 1

table, root = newton_raphson(f, df, 1.5)

pd.set_option('display.float_format','{:8.4f}'.format)
pd.set_option('display.colheader_justify','center')

print(table)
print("\nApproximate Root:", root)
```
