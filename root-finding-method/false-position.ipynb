```python
# import library
import pandas as pd

# define function
def false_position(f, a, b, tol, max_iter):
    
    # check whether the initial interval is valid
    if f(a)*f(b)>=0:
        return "Invalid Interval"

    # create an empty list to store iteration data
    data = []
    
    # initialize the previous approximation
    old_c = None

    # start the iteration loop
    for i in range(1, max_iter+1):
        
        # calculate the new approximation using false-position formula
        c = (a*f(b)-b*f(a))/ (f(b)-f(a))

        # calculate relative error
        if old_c is None:
            error = None
        else:
            error = abs((c-old_c)/c)

        # store the iteration values in the data list
        data.append([i,a,b,c,f(c),error])

        # check the convergence condition
        if error is not None and error<tol:
            break
        
        # update the interval according to the sign of f(a) and f(b)
        if f(a)*f(c)<0:
            b = c
        else:
            a = c
        
        # update the previous approximation
        old_c = c

    # convert iteration data into pandas dataframe
    df = pd.DataFrame(data, columns = ["Iteration","a","b","c","f(c)","Relative Error"])

    # return the dataframe and approximation root
    return df, c

# define the given function
f = lambda x: x**3 - x - 2

# call the function with inital interval [1,2], tolerance maximum iteration
df, root = false_position(f,1,2,1e-6,100)

pd.set_option('display.colheader_justify','center')
pd.set_option('display.float_format','{:8.4f}'.format)


# display the iteration table and approximate root
print(df)
print("\nApproximadte root:", root)     
```
