```python
import pandas as pd

def iterative_method(g, x0, tol=0.0001, max_iter=100):

    data = []

    for i in range(1, max_iter + 1):

        x1 = g(x0)

        error = abs(x1 - x0)

        data.append([i, x0, x1, error])

        if error < tol:
            break

        x0 = x1

    table = pd.DataFrame(data, columns=[
        "Iteration", "x0", "x1", "Error"
    ])

    return table, x1


g = lambda x: 1 / (1 + x**2)

table, root = iterative_method(g, 0.5)

pd.set_option('display.colheader_justify','center')
pd.set_option('display.float_format','{:10.6f}'.format)

print(table)
print("\nApproximate Root:", root)
```
