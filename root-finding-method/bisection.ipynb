### **Algorithm of Bisection Method**

Step 1: Define the function $f(x)$, choose the initial interval $[a,b]$, and set the tolerance $e$.

Step 2: Check if $f(a)\times f(b)\ge0$. If true, the initial gueses do not bracket a root.

Step 3: Calculate the midpoint $c = \frac{a+b}{2}$.

Step 4: Check if $f(c)= 0$ or if the interval width $\frac{(b-a)}{2}<e$. If true, $c$ is the root. Go to step 7.

Step 5: Check if $f(a)\times f(c)<0$. If true,set the new upper bound $b=c$. Otherwise, set the new lower bound $a=c$.

Step 6: Repeat step 3 to step 5 until the condition in step 4 met.

Step 7: Display $c$ as the the final root and stop.

  


```python
# import library

import time
import math

# define function

def bisection(f, a:float, b:float, tol:float = 1e-4, max_iter = 20):

    # ensure the interval is in ascending order

    if a > b :
        a, b = b, a    # initial a and b

    # count function evaluation

    func_eval = 0

    fa = f(a)
    func_eval+=1

    fb = f(b)
    func_eval+=1

    if fa*fb >= 0:

        print("Wrong interval! Roots does not exist in this range.")
        return (None, 0, False, "Invalid Interval", func_eval, None)


    # table format

    width = 15

    print()     # blank line

    print(
        f"{'Iteration':<{width}}"
        f"{'a':^{width}}"
        f"{'b':^{width}}"
        f"{'c':^{width}}"
        f"{'f(c)':^{width}}"
        f"{'error(%)':>{width}}"
    )

    print("_"*90)

    # variable

    previous_c = None
    converged = False
    stopping_reason = "Maximum iteration Reached"
    final_residual = None

    # iteration loop

    for iteration in range (1,max_iter+1):

        # midpoint

        c = (a+b)/2

        # evaluate function at midpoint

        fc = f(c)
        func_eval+=1

        if previous_c is None:
            error = "-"

        else:
            error = abs((c-previous_c)/c)*100

        # print table

        print(
            f"{iteration:^10}"
            f"{a:>15.4f}"
            f"{b:>15.4f}"
            f"{c:>15.4f}"
            f"{fc:>15.4f}"
            f"{error if error=='-' else f'{error:.8f}':>20}"
        )

        # stopping criterion 1

        if abs(fc) < tol :

            converged = True
            stopping_reason = "f(c) is less than tolerance"
            final_residual = abs(fc)
            return (c, iteration, converged, stopping_reason, func_eval, final_residual)

        # update interval

        if fa*fc < 0:
            b = c
            fb = fc

        else:
            a = c
            fa = fc

        previous_c = c

        # stoping critarion 2

        if (b-a)/2 < tol:

            converged = True
            stopping_reason = "(b-a)/2 is less than tolerance"
            final_residual = abs(fc)
            return (c, iteration, converged, stopping_reason, func_eval, final_residual)


    # store the residual after reaching the maximum iterations

    final_residual = abs(fc)

    return (c, iteration, converged, stopping_reason, func_eval, final_residual)

# function example

def f(x):
    return x - 0.8 - 0.2*math.sin(x)

# count execution time

start = time.perf_counter()

root, iteration, converged, stopping_reason, func_eval, final_residual = bisection(f, 0.0, math.pi/2)

end = time.perf_counter()

execution_time = end - start


# root summary

if root is not None:

    print("_"*90)

    print("\nROOT SUMMARY")

    print("_"*12)

    print()

    print(f"{'Method':<25}: Bisection Method")

    print(f"{'Initial Interval':<25}: [0.0, pi/2]")

    print(f"{'Approximate Root':<25}: {root:.8f}")

    print(f"{'f(root)':<25}: {f(root):.8e}")

    print(f"{'Final Residual':<25}: {final_residual:.8e}")

    print(f"{'Iteration':<25}: {iteration}")

    print(f"{'Function Evaluations':<25}: {func_eval}")

    print(f"{'Converged':<25}: {'Yes' if converged else 'No'}")

    print(f"{'Stopping Reason':<25}: {stopping_reason}")

    print(f"{'Execution Time':<25}: {execution_time:.8e} seconds")
```
