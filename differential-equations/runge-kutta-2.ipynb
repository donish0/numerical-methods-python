```python
from math import sin, cos

# define the differential equation
def f(x, y):
    return sin(x) + y * cos(x)

# initial conditions and parameters
x = 0.0
y = 1.0
h = 0.1       # step size
n = 10        # number of iterations

# display table header
print(f"{'Iteration':^15}{'x':^15}{'y':^15}")
print("-" * 45)

# apply RK2 (Heun's method)
for i in range(1, n + 1):
    k1 = f(x, y)                   # slope at the beginning
    k2 = f(x + h, y + h * k1)      # slope at the end

    y = y + (h / 2) * (k1 + k2)    # update y
    x = x + h                      # update x

    print(f"{i:^15}{x:^15.4f}{y:^15.6f}")
```
