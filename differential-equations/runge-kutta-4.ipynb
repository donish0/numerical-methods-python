```python
# define the differential equation
def f(x, y):
    return x**2 + y**2

# initial conditions and parameters
x = 0.0
y = 0.5
h = 0.1       # step size
n = 10        # number of iterations

# display table header
print(f"{'Iteration':^15}{'x':^15}{'y':^15}")
print("-" * 45)

# apply RK4 method
for i in range(1, n + 1):
    k1 = f(x, y)
    k2 = f(x + h/2, y + h*k1/2)
    k3 = f(x + h/2, y + h*k2/2)
    k4 = f(x + h, y + h*k3)

    y = y + (h/6) * (k1 + 2*k2 + 2*k3 + k4)  # update y
    x = x + h                                # update x

    print(f"{i:^15}{x:^15.4f}{y:^15.6f}")
```
