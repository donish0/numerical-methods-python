```python
# define the differential equation
def f(x, y):
    return 2*x + y

# initial conditions and parameters
x0 = 0.0
y0 = 1.0
h = 0.1       # step size
xn = 1.0      # final value of x

# initialize x and y values
n = round((xn - x0) / h)
x = [x0 + i*h for i in range(n + 1)]
y = [0] * (n + 1)
y[0] = y0

# apply RK4 for the first three steps
for i in range(3):
    k1 = h * f(x[i], y[i])
    k2 = h * f(x[i] + h/2, y[i] + k1/2)
    k3 = h * f(x[i] + h/2, y[i] + k2/2)
    k4 = h * f(x[i] + h, y[i] + k3)

    y[i+1] = y[i] + (k1 + 2*k2 + 2*k3 + k4) / 6

# display table header
print(f"{'Iteration':^15}{'x':^15}{'yp':^15}{'yc':^15}")
print("-" * 60)

# apply Adams-Bashforth-Moulton method
for i in range(3, n):

    # predictor (Adams-Bashforth)
    yp = y[i] + h/24 * (55*f(x[i], y[i])
                        - 59*f(x[i-1], y[i-1])
                        + 37*f(x[i-2], y[i-2])
                        - 9*f(x[i-3], y[i-3]))

    # corrector (Adams-Moulton)
    yc = y[i] + h/24 * (9*f(x[i+1], yp)
                        + 19*f(x[i], y[i])
                        - 5*f(x[i-1], y[i-1])
                        + f(x[i-2], y[i-2]))

    y[i+1] = yc  # update y

    print(f"{i+1:^15}{x[i+1]:^15.4f}{yp:^15.6f}{yc:^15.6f}")
```
