###**Newton-Raphson Method**

```python
def f(x):
    return x**3 - x - 2

def df(x):
    return 3*x**2 - 1

def newton_raphson(x0, tol=1e-6, max_iter=100 ):
    i = 0
    while i < max_iter:
        x1 = x0 - f(x0) / df(x0)
        i += 1
        print(f"Iteration {i}: x = {x1:.3f}")
        if abs(x1 - x0) < tol:
            break
        x0 = x1
    print("\nRoot =", f"{x1:.3f}")
    print("Iterations =", i)

newton_raphson(1.5)
```
