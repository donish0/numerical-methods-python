###**Simpson's 3/8 Rule**

```python
# import library
import sympy as sp

# define variable
x = sp.symbols('x')

# define function
f = 0.2 + 25*x - 200*x**2 + 675*x**3 - 900*x**4 + 400*x**5

# given values
a = 0.0
b = 0.8
n = 21    # n must be multiple of 3

# stepsize
h = (b-a)/n

# simpson's 3/8 rule
simpson_38 = (3*h/8)*(f.subs(x,a) + f.subs(x,b) + 3*sum(f.subs(x, a + i*h)for i in range(1,n)if i % 3 != 0) + 2*sum(f.subs(x, a + i*h) for i in range (3,n,3)))

# exact value
exact = sp.integrate(f,(x,a,b))

# absoluete error
abs_error = abs(exact - simpson_38)

# relative error
rel_error = (abs_error/abs(exact))*100

#second derivative
f4 = sp.diff(f,x,4)

# average value of second derivative
avg_f4 = (1/b-a)*sp.integrate(f4,(x,a,b))

# truncation error
trunc_error = -((b-a)/80)*h**4*avg_f4

# display result
print(f"\napproximate value = {simpson_38:.6f}")
print(f"\nexact value = {exact:.6f}")
print(f"\nabsolute error = {abs_error:.6f}")
print(f"\ntruncation error = {trunc_error:.6f}")
```
