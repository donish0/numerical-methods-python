####LU Decompositon Method

```python
#import library
import numpy as np
from scipy.linalg import lu_factor, lu_solve

A = np.array([[2,1,1],
             [3,2,1],
             [2,1,2]], dtype = float)

b = np.array([8,13,9], dtype = float)

# LU factorization
lu, piv = lu_factor(A)
x = lu_solve((lu,piv),b)

print("solution using LU decompostion:")
print(x)
```
