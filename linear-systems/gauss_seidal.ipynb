```python
import numpy as np

A = np.array([[10,-1,2],
             [-1,11,-1],
             [2,-1,10]], dtype = float)

b = np.array([6,25,-11], dtype = float)

n = len(b)

x = np.zeros(n)
x_old = np.zeros(n)

tol = 1e-6
max_iter = 100

for k in range(max_iter):

    x_old = x.copy()

    for i in range(n):

        sum_value = 0.0

        for j in range(n):
            if i != j :
                sum_value+= A[i,j]*x[j]

        x[i] = (b[i] - sum_value)/A[i,i]

    if np.max(np.abs(x - x_old)) < tol:
        break

print(x)
```
