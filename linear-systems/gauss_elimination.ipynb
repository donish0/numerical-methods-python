```python
import numpy as np

A = np.array([[1,2,3,],
             [4,5,6],
            [7,7,8]], dtype = float)

b = np.array([1,0,0], dtype = float)

Aug = np.column_stack((A,b))

n = len(b)

#forward elimination

for i in range(n-1):
    for j in range(i+1, n):

        factor = Aug[j,i]/Aug[i,i]
        Aug[j] = Aug[j]- factor*Aug[i]

    print("step",i+1)
    print(Aug)

# back substitution

x = np.zeros(n)

for i in range(n-1, -1, -1):
    sum_value = 0.0

    for j in range(i+1, n):
        sum_value+=Aug[i,j]*x[j]

    x[i] = (Aug[i,n]-sum_value)/Aug[i,i]

print(x)
```
