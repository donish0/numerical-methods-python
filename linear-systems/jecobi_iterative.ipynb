```python
import numpy as np

A = np.array([[10,-1,2],
             [-1,11,-1],
             [2,-1,10]], dtype = float)

b = np.array([6,25,-11], dtype = float)

n = len(b)

x = np.zeros(n)
x_new = np.zeros(n)

tol = 1e-6
max_iter = 100

for k in range(max_iter):
    # 1 k-loop execution = 1 iteration

    for i in range(n):
        # 1 i-loop execution = 1 equation is calculated

        sum_value = 0.0

        for j in range(n):
            # 1 j-loop execution = 1 A[i,j]*x[j] term is calculated

            if i != j :
                sum_value+= A[i,j]*x[j]

        x_new[i] = (b[i] - sum_value)/A[i,i]

    if np.max(np.abs(x_new - x)) < tol:
        break
    x = x_new.copy()

print(x_new)
```
