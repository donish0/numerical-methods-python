```python
import numpy as np

A = np.array([[1,1/2,1/3],
             [1/2,1/3,1/4],
             [1/3,1/4,1/5]], dtype = float)

b = np.array([1,0,0])

Aug = np.column_stack((A,b))

n = len(b)

for i in range(n):

    Aug[i] = Aug[i]/Aug[i,i]

    for j in range(n):

        if i !=j:
            Aug[j] = Aug[j]- Aug[j,i]*Aug[i]

    print(f"step{i+1}")
    print(Aug)

x = Aug[0:3,3] # x = Aug[:,-1]

print(x)
```
