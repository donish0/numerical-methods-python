```python

# import library
import numpy as np                     # for numerical calculation
import pandas as pd                    # for displaying iteration results
from IPython.display import display    # for formatted table output

# define the matrix
A = np.array([[1,3,-1],
            [3,2,4],
            [-1,4,10]], dtype = float)

# convergence tolerance
tol = 1e-6

# initial eigenvector guess
x = np.array([1,1,1], dtype = float)

# previous eigenvalue
lambda_old = 0

# initial error
error = 1

# iteration counter
i = 0

# store iteration results
data = []

# repeat until convergence
while error > tol:

    # increase iteration  counter
    i+=1

    # matrix-vector multiplication
    y = A @ x

    # Normalize the eigenvector
    x = y/np.linalg.norm(y)

    # estimate the eigenvalue using Rayleigh quotient
    lambda_new = x @ (A @ x)

    # calculate the absolute error
    error = abs(lambda_new - lambda_old)

    # update previous eigenvalue
    lambda_old = lambda_new

    # format eigenvector to six decimal places
    eigenvector = "["+ ", ".join(f"{v:.6f}" for v in x) +"]"

    # save iteration results
    data.append([i, lambda_new, eigenvector, error])

# create a DataFrame
df = pd.DataFrame(data, columns = ["Iteration", "Eigenvalue", "Eigenvector", "Error"])

# display a formatted and center-aligned table
display(
    df.style
    .format({"Eigenvalue": "{:.6f}",                 # format eigenvalue
             "Error": "{:.6f}"                       # format error
    })
    .set_properties(**{"text-align": "center"})      # center cell values
    .set_table_styles([
        {"selector": "th", "props": [("text-align", "center")]}     # center headers
    ])
)

# display final results
print("\nMaximum Eigenvalue =", round(lambda_new,4))
print("Corresponding Eigenvector =", np.round(x,4).tolist())        # convert a numpy array to a list for comma-separated output
```
