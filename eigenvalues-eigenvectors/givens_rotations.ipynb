```python

# import library
import numpy as np

# input symmetric matrix
A = np.array([[ 6.0,  2.0,  1.0,  3.0,  2.0,  1.0],
              [ 2.0,  7.0,  2.0,  1.0,  3.0,  2.0],
              [ 1.0,  2.0,  8.0,  2.0,  1.0,  3.0],
              [ 3.0,  1.0,  2.0,  9.0,  2.0,  1.0],
              [ 2.0,  3.0,  1.0,  2.0, 10.0,  2.0],
              [ 1.0,  2.0,  3.0,  1.0,  2.0, 11.0]])

tol = 1e-6          # set tolerance
n = len(A)          # satrix size
T = A.copy()        # copy original matrix
Q = np.eye(n)       # initialize orthogonal matrix
i = 0               # initialize iteration counter

for k in range(n-2):                   # loop through columns
    for r in range(n-1, k+1, -1):      # loop through rows from bottom
        if abs(T[r, k]) > tol:         # check nonzero element
            i += 1                     # count rotation
            a = T[k+1, k]              # first element
            b = T[r, k]                # element to eliminate

            radius = np.sqrt(a**2 + b**2)    # compute magnitude
            c = a/radius                     # compute cosine
            s = b/radius                     # compute sine

            G = np.eye(n)             # initialize rotation matrix
            G[k+1, k+1] = c           # set cosine
            G[k+1, r] = -s            # set negative sine
            G[r, k+1] = s             # set sine
            G[r, r] = c               # set cosine

            T = G.T @ T @ G           # apply similarity transformation
            Q = Q @ G                 # accumulate rotations

            print(f"\nIteration : {i}")
            print(f"{'_' * 65}\n")
            print("Rotation Matrix, G:\n")
            print(np.round(G, 6))
            print("\nTransformed Matrix, T:\n")
            print(np.round(T, 6))
            print("\n")

eigenvalues, eigenvectors_T = np.linalg.eigh(T)  # Compute eigenpairs

eigenvectors = Q @ eigenvectors_T  # Recover original eigenvectors

print(f"{"_"* 65}\n")
print("FINAL RESULTS")
print("_"* 65)
print("\nTridiagonal Matrix, T:\n")
print(np.round(T, 6))
print("\nEigenvalues:\n")
print(np.round(eigenvalues, 6).tolist())
print("\nEigenvectors as columns:\n")
print(np.round(eigenvectors, 6))
```
