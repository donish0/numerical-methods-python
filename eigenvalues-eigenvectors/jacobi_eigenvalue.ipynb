```python

# import library for numerical computation
import numpy as np

# define the symmetric matrix
A = np.array([[5.0, 1.0, 0.0],
              [1.0, 4.0, 1.0],
              [0.0, 1.0, 3.0]])

tol = 1e-6            # convergence tolerance
n = len(A)            # matrix size
V = np.eye(n)         # initialize eigenvector matrix
error = 1.0           # initial error
iter = 0                 # iteration counter

# repeat until convergence
while error > tol:
    error = 0.0       # reset maximum error

    # find the largest off-diagonal element
    for r in range(n):
        for q in range(r + 1, n):
            if abs(A[r, q]) > error:
                error = abs(A[r, q])    # update maximum error
                i = r                   # row index
                j = q                   # column index

    if error <= tol:
        break         # stop if converged

    iter += 1            # count iteration

    # calculate rotation angle
    theta = 0.5 * np.arctan2(2 * A[i, j], A[i, i] - A[j, j])
    c = np.cos(theta)
    s = np.sin(theta)

    # Construct rotation matrix
    R = np.eye(n)     # start with identity matrix
    R[i, i] = c       # set diagonal entry
    R[i, j] = -s      # set upper entry
    R[j, i] = s       # set lower entry
    R[j, j] = c       # set diagonal entry

    A = R.T @ A @ R   # apply Jacobi rotation
    V = V @ R         # update eigenvectors

    # display iteration results
    print(f"Iteration: {iter}")
    print("_"*50)
    print("\nA =\n")
    print(np.round(A, 6))
    print("\n")

# extract eigenvalues and eigenvectors
eigenvalues = np.diag(A)  # diagonal elements
eigenvectors = V

# display final results
print("_"*50)
print("\nFinal Result:")
print("_"*50)
print("\nEigen Values =", np.round(eigenvalues, 6).tolist())
print("\nEigenvectors =\n")
print(np.round(eigenvectors, 6))
```
