###**Fixed-Point Iteration Method**


```python
def g(x):
    return (x + 2)**(1/3)     # x = g(x)

def fixed_point(x0, tol=1e-6, max_iter=100):
    i = 0

    while i < max_iter:
        x1 = g(x0)
        i += 1
        print(f"Iteration {i}: x = {x1:.3f}")
        if abs(x1 - x0) < tol:
            break
        x0 = x1

    print("\nRoot =", f"{x1:.3f}")
    print("Iterations =", i)

fixed_point(1)
```
