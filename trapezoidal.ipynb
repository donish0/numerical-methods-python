###**Composite Trapezoidal Rule**

```python
import sympy as sp

# define variable
x = sp.symbols('x')

# define function
f = 0.2 + 25*x - 200*x**2 + 675*x**3 - 900*x**4 + 400*x**5

# given values
a = 0.0
b = 0.8
n = 20

# stepsize
h = (b-a)/n

# composite trapezoidal rule
trapezoidal = (h/2)*(f.subs(x,a) + f.subs(x,b) + 2*sum(f.subs(x, a + i*h)for i in range(1,n)))

# exact value
exact = sp.integrate(f,(x,a,b))

# absoluete error
abs_error = abs(exact - trapezoidal)

# relative error
rel_error = (abs_error/abs(exact))*100

#second derivative
f2 = sp.diff(f,x,2)

# average value of second derivative
avg_f2 = (1/b-a)*sp.integrate(f2,(x,a,b))

# truncation error
trunc_error = -((b-a)/12)*h**2*avg_f2

# display result
print(f"approximate value = {trapezoidal:.6f}")
print(f"\nexact value = {exact:.6f}")
print(f"\nabsolute error = {abs_error:.6f}")
print(f"\ntruncation error = {trunc_error:.6f}")
```
