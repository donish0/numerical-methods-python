###**Simpson's 1/3 Rule**

```python
# import library
import sympy as sp

# define variable
x = sp.symbols('x')

# define function
f = 0.2 + 25*x - 200*x**2 + 675*x**3 - 900*x**4 + 400*x**5

# given values
a = 0.0
b = 0.8
n = 20     # n must be even number

# stepsize
h = (b-a)/n

# simpson's 1/3 rule
simpson_13 = (h/3)*(f.subs(x,a) + f.subs(x,b) + 4*sum(f.subs(x, a + i*h)for i in range(1,n,2)) + 2*sum(f.subs(x, a + i*h) for i in range (2,n,2)))

# exact value
exact = sp.integrate(f,(x,a,b))

# absoluete error
abs_error = abs(exact - simpson_13)

# relative error
rel_error = (abs_error/abs(exact))*100

#second derivative
f4 = sp.diff(f,x,4)

# average value of second derivative
avg_f4 = (1/b-a)*sp.integrate(f4,(x,a,b))

# truncation error
trunc_error = -((b-a)/180)*h**4*avg_f4

# display result
print(f"approximate value = {simpson_13:.6f}")
print(f"\nexact value = {exact:.6f}")
print(f"\nabsolute error = {abs_error:.6f}")
print(f"\ntruncation error = {trunc_error:.6f}")
```
