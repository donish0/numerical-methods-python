###**False-Position Method**

```python
def f(x):
    return x**3 - x - 2

def false_position(a, b, tol=1e-6, max_iter=100):

    if f(a) * f(b) >= 0:
        print("No root in the given interval.")
        return

    i = 0

    while i < max_iter:

        c = a - f(a) * (b - a) / (f(b) - f(a))
        i += 1
        print(f"Iteration {i}: c = {c:.3f}, f(c) = {f(c):.3f}")

        if abs(f(c)) < tol:
            break
        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

    print("\nRoot =", f"{c:.3f}")
    print("Iterations =", i)

false_position(1, 2)
```
