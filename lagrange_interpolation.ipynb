###**Lagrange Interpolation**

```python
def lagrange(x_data, y_data, x):

    n = len(x_data)
    result = 0

    for i in range(n):
        term = y_data[i]

        for j in range(n):
            if i != j:

                term = term * (x - x_data[j]) / (x_data[i] - x_data[j])
        result += term

    return result

x_data = [1, 2, 3, 4]
y_data = [1, 4, 9, 16]
x = 2.5

answer = lagrange(x_data, y_data, x)

print("Interpolated value =", f"{answer:.6f}")
```
